```python
# Prob 3
from random import sample
from math import perm, comb
from itertools import permutations, combinations
```

```python
# Prob 3 Formula Aranjamente
def aranjamente(str, k, numar_total=False, aleator=False):
    if numar_total:
        return perm(len(str),k)
    elif aleator:
        return sample(str,k)
    else:
        rez = []
        combs = combinations(str, k)
        for combi in combs:
            rez += list(permutations(combi))
        return rez
```

```python
# Prob 3.1
aranjamente('word',2)
```

```python
# Prob 3.2
aranjamente('word',2, numar_total=True)
```

```python
# Prob 3.3
aranjamente('word',2, aleator=True)
```

```python
# Prob 3 Formula Combinari
def combinari(str, k, numar_total=False, aleator=False):
    if numar_total:
        return comb(len(str),k)
    elif aleator:
        al_sample = sample(str,k)
        al_sample.sort(key = lambda x: str.index(x))
        return al_sample
    else:
        return list(combinations(str,k))
```

```python
# Prob 3.4
combinari('word',2)
```

```python
# Prob 3.5
combinari('word',2, numar_total=True)
```

```python
# Prob3.6
combinari('word',2, aleator=True)
```
