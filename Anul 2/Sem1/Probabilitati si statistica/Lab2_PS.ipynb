```python
from itertools import product
from random import randint
from matplotlib.pyplot import plot, grid, xticks, yticks, title, show

data = {}
for i in range(3, 19):
  data[i] = 0

for i in range(100_000):
  z = [randint(1, 6) for _ in range(3)]
  data[sum(z)] += 1
print(data)
```

```python
for i in range(100_000_00):
    z = [randint(1, 6) for _ in range(3)]
    data[sum(z)] += 1
plot(data.keys(), data.values(), "*")
```

```python
count = { s:0 for s in range(3,19)}
for z in product (range(1,7), repeat=3):
    count[sum(z)] += 1
plot(count.keys(), count.values(), "*")
xticks(range(3,19))
grid()
show()
```

```python
from matplotlib.pyplot import axis, plot, show
from random import random
from math import dist

axis('square')
axis((0, 1, 0, 1))

for i in range(1000):
  F=[random(),random()]
  E = [0.5, 0.5]
  if dist(F, E) < 0.5:
    plot(F[0],F[1],'ro')
  else:
    plot(F[0],F[1],'go')
show()
random()
```

```python
from matplotlib.pyplot import axis, plot, show
from random import random
from math import dist

axis('square')
axis((0, 1, 0, 1))

N = 1000
k = 0
E = [0.5, 0.5]
for i in range(N):
    F = [random(), random()]
    if dist(F, E) < 0.5:
        plot(F[0], F[1], 'ro')
        k += 1
    else:
        plot(F[0], F[1], 'go')
show()

print(4*(k / N))
```

```python
from matplotlib.pyplot import axis, plot, show
from random import random
from math import dist

axis('square')
axis((0, 1, 0, 1))
cnt = 0
for i in range(1000):
  F=[random(),random()]
  E = [0.5, 0.5]
  if dist(F, E) < dist(F, [1,0]) and dist(F, E) < dist(F, [1,1]) and dist(F, E) < dist(F, [0,0]) and dist(F, E) < dist(F, [0,1]):
    cnt += 1
    plot(F[0],F[1],'ro')
  else:
    plot(F[0],F[1],'go')
print((cnt/1000) * 4)
show()
random()
```

```python
from matplotlib.pyplot import axis, plot, show
from random import random
from math import dist

axis('square')
axis((0, 1, 0, 1))
cnt = 0
N = 1000
for i in range(N):
  F=[random(),random()]
  E = [0.5, 0.5]
  if dist(F, E) < F[1]:
    cnt += 1
    plot(F[0],F[1],'ro')
  else:
    plot(F[0],F[1],'go')
print(cnt/N)
show()
```
