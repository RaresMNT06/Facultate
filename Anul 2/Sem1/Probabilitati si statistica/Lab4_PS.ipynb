```python
from scipy.stats import bernoulli, binom
from matplotlib.pyplot import bar, show, hist, grid, legend, xticks
```

```python
def mers_aleatoriu(n, p):
  poz = 0
  for i in range(n):
    direction = bernoulli.rvs(p)
    if direction == 1:
      poz += 1
    else:
      poz -= 1
  return poz
data = [mers_aleatoriu(10, 0.5) for i in range(1000)]
bin_edges = [i+0.5 for i in range(-10, 11)]
hist(data, bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',
     alpha = 0.5, label = 'frecvente relative')
xticks(range(-10, 11))
show()
```

```python
def mers_aleatoriu(n, p):
  poz = 0
  for i in range(n):
    direction = bernoulli.rvs(p)
    poz = poz + (direction * 2 - 1)
  return poz
data = [mers_aleatoriu(10, 0.5) for i in range(1000)]
bin_edges = [i+0.5 for i in range(-10, 11)]
hist(data, bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',
     alpha = 0.5, label = 'frecvente relative')
xticks(range(-10, 11))

n = 10
p = 0.5
distribution = dict([(2*k-n, binom.pmf(k, n, p)) for k in range(0, n+1)])
bar(distribution.keys(), distribution.values(), width = 0.85, color = 'red', edgecolor = 'black',
    alpha = 0.6, label = 'probabilitati')
show()
```
